## Задание 1: Зеркальное отражение цифр

Пользователь вводит пятизначное число. Программа должна зеркально отразить три центральные цифры. Первая и последняя остаются на местах.

<br>

**Примеры:**
* `23456 -> 25436`
* `30789 -> 38709`

```python
user_input = [23456, 30789, 4786125]

def reverse_center(number):
    number = str(number)
    return int(number[0] + number[1:-1][::-1] + number[-1])

result = [print(input, '->', reverse_center(input)) for input in user_input]
```

## Задание 2: Подсчет выходных до отпуска

Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.  

<br>

**Примеры:**
* `4  -> 0`
* `6  -> 1`
* `14 -> 4`

```python
from datetime import date, timedelta
from dateutil.relativedelta import relativedelta

user_input = [-10, 0, 1, 4, 6, 14, 20, 51]

first_vacation_date = date(2026, 1, 4)

def weekends_before_vacation(days, vacation_date):
    if days <= 1:
        return "incorrect input"

    from_date = vacation_date - timedelta(days)

    weeks = days // 7

    from_date_week_weekends = (2 - from_date.isoweekday() // 6 - from_date.isoweekday() // 7) if weeks > 0 else 0    
    middle_weeks_weekends = 2 * (weeks - 1) if weeks > 1 else 0 
    vacation_week_weekends = vacation_date.isoweekday() // 7

    return from_date_week_weekends + middle_weeks_weekends + vacation_week_weekends #, date


result = [print(input, '->', weekends_before_vacation(input, first_vacation_date)) for input in user_input]
```

## Задание 3: Плитка шоколада

Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

<br>

**Примеры:**
* `3, 4, 6  -> True`
* `5, 7, 8  -> False`
* `4, 5, 12 -> True`

```python
user_input = [
    [3, 4, 6],
    [5, 7, 8],
    [4, 5, 12],
    [3, 5, 10],
    [5, 11, 19]
]

def is_separable(input):
    side1, side2, pieces = input

    if pieces % side1 != 0 and pieces % side2 != 0:
        return False
    
    if pieces % side2 == 0:
        side1, side2 = side2, side1

    return pieces // side1 < side2

result = [print(f"{input[0]}, {input[1]}, {input[-1]} -> {is_separable(input)}") for input in user_input]
```

## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
user_input = [3, 15, 234, 143, 3999]
roman_digits = {
    1: "I", 4: "IV", 5: "V",
    9: "IX", 10: "X",
    40: "XL", 50: "L",
    90: "XC", 100: "C",
    400: "CD", 500: "D",
    900: "CM", 1000: "M"
}

def to_roman_digits(number):
    roman_number = ""
    for key in reversed(roman_digits.keys()):
        times = number // key
        if times > 0:
            roman_number += times * roman_digits[key]
            number -= times * key
    
    return roman_number

result = [print(input, '->', to_roman_digits(input)) for input in user_input]
```
