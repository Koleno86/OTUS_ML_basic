1. Написать функцию, которая принимает неограниченное количество чисел в виде позиционных аргументов и ключевой аргумент — операцию над этими числами (сложение или умножение).
Функция должна возвращать результат выполнения указанной операции.

```python
import operator

def action(*numbers, operation=operator.add):
    result = numbers[0]
    for number in numbers[1:]:
        result = operation(result, number)
    
    return result

print("Multipication:", action(3, 4, 1, 3, 55.8, 445, operation=operator.mul))
print("Addition:", action(3, 4, 1, 3, 55.8, 445, operation=operator.add))
```

2. Написать функцию для ввода из консоли целого числа.
Если введено не число, функция должна вывести соответствующее сообщение и предложить ввести значение заново — до тех пор, пока пользователь не введёт корректное число.

```python
def input_number():
    while number := input("Введите правильное число: "):
        try:
            number = int(number)

            if number == 400:
                print("Поздравляю, наконец-то угадали! :)")
                break
        
            print(number, "Неверное число, попробуйте еще раз")
        except ValueError:
            print(number, "Вы ввели не целое число, попробуйте еще раз")

input_number()
```

3. Написать функцию, которая создаёт абсолютный путь к файлу.

Позиционные аргументы:

название диска,

неограниченное количество папок,

имя файла (без расширения).


Ключевые аргументы:
ext — расширение файла,

sep — разделитель (по умолчанию '/').


Пример:
full_path('c:', 'work', 'python', 'function', 'main', ext='py') ➜ 'c:/work/python/function/main.py'

```python
def make_full_path(*elements, sep='/', ext=''):
    ext = ext.strip()
    sep = sep.strip()
    return f"{sep.join(elements)}{'.' if ext else ''}{ext}"

print(make_full_path('c:', 'work', 'python', 'function', 'main', ext='py'))
```


4. Написать функцию, которая принимает список, состоящий из объектов разных типов, и возвращает словарь, где:
ключи — типы данных объектов;

значения — списки объектов соответствующего типа.

```python
from pprint import pprint

user_input = [3, 1, 6, 4, 66, "ff", "ewr", "3", "66", "ff", "!", 66, 12.0, ".4d"]

def types_info(list_):
    result = {}
    for element in list_:
        type_ = type(element).__name__
        result.setdefault(type_, set())
        result[type_].add(element)

    return {type_: sorted(list(result[type_])) for type_ in result}

pprint(types_info(user_input))
```
