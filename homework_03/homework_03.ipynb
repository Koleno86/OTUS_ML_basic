## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
import re
from enum import StrEnum
from typing import Callable, Self, Dict

class Case(StrEnum):
    AUTO = "auto"
    SNAKE = r'^[a-z][a-z0-9]*(_[a-z0-9]+)*$'
    #CAMEL = r'^[a-z][a-zA-Z0-9]*$'
    PASCAL = r'^[A-Z][a-zA-Z0-9]*$'

    @staticmethod
    def detect(text: str) -> Self:
        cases = list(Case)
        for case in cases[1:]:
            if re.match(case, text):
                return case
        
        return Case.SNAKE
    
    @staticmethod
    def next(case: Self) -> Self:
        cases = list(Case)
        current_index = cases.index(case)
        return cases[current_index % (len(cases) - 1) + 1]


CASE_REGISTRY: Dict[Case, Callable[[str], str]] = {} 

def register(case: Case):
    def decorator(func: Callable[[str], str]) -> Callable[[str], str]:
        CASE_REGISTRY[case] = func
        return func
    return decorator

@register(Case.SNAKE)
def to_snake_case(text: str) -> str:
    text_ = re.sub(r'(?<!^)(?=[A-Z])', '_', text)
    text_ = re.sub(r'[^a-zA-Z0-9]', '_', text_)
    return text_.lower()

@register(Case.PASCAL)
def to_pascal_case(text: str) -> str:
    words = re.split(r'[^a-zA-Z0-9]+', text)
    return "".join([word[0].upper() + word[1:] for word in words if word])

#@register(Case.CAMEL)
def to_camel_case(text: str) -> str:
    to_pascal_case = CASE_REGISTRY.get(Case.PASCAL)
    if to_pascal_case:
        text = to_pascal_case(text)
        return text[0].lower() + text[1:]

def swap_case(text: str, target_case: Case = Case.AUTO) -> str:
    if target_case == Case.AUTO:
        current_case = Case.detect(text)
        target_case = Case.next(current_case)

    to_case_fn = CASE_REGISTRY.get(target_case)
    if to_case_fn:
        return to_case_fn(text)
  

user_input = ["otus_course", "PythonIsTheBest", "hAllO_Python"]

print("Auto")
for input in user_input:
    print(input, '->', swap_case(input, Case.AUTO))

print()
print("To snake case")
for input in user_input:
    print(input, '->', swap_case(input, Case.SNAKE))

print()
print("To pascal case")
for input in user_input:
    print(input, '->', swap_case(input, Case.PASCAL))    
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
from dateutil.parser import parse, ParserError

user_input = ['29.02.2000', '29.02.2001', '31.04.1962']

def is_valid(date_str):
    try:
        parse(timestr=date_str)
        return True
    except ParserError:
        return False

for input in user_input:
    print(input, '->', is_valid(input))
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
from dataclasses import dataclass, field
from typing import Callable, Self, Tuple, List
from functools import wraps

def unique_id(cls):
    existing_ids = set()
    original_init = cls.__init__

    @wraps(original_init)
    def new_init(self, *args, **kwargs):
        id = kwargs.get('id') or (args[0] if args else None)
        if id in existing_ids:
            raise ValueError(f"ID '{id}' уже используется")
        original_init(self, *args, **kwargs)
        existing_ids.add(id)
    cls.__init__ = new_init
    return cls

@unique_id
@dataclass
class User:
    id: str
    firstname: str
    lastname: str
    age: str

    def __post_init__(self):
        self.id = self.id.zfill(8)
        self.firstname = self.firstname.capitalize()
        self.lastname = self.lastname.capitalize()

    def __len__(self):
        return len(self.id)


class Predicate:
    def __init__(self, fn: Callable[[User], bool]):
        self.fn = fn

    def __call__(self, user: User) -> Tuple[bool, str]:
        is_valid, error = self.fn(user)
        return is_valid, error if not is_valid else ""

    def __and__(self, other: Self) -> Self:
        def combined_fn(user: User) -> Tuple[bool, str]:
            valid1, error1 = self(user)
            valid2, error2 = other(user)

            errors = [] 
            if not valid1:
                errors.append(error1)
            if not valid2:
                errors.append(error2)

            return valid1 and valid2, ". ".join(errors)

        return Predicate(combined_fn)

    
is_id = Predicate(lambda user: (len(user) <= 8 and user.id.isdigit(), 
                                "ID должно быть меньше 8 цифр и целым числом"))       
is_name = Predicate(lambda user: (user.firstname.isalpha() and user.lastname.isalpha(), 
                                  "В имени есть недопустимые символы"))
is_age = Predicate(lambda user: (user.age.isdigit() and 18 <= int(user.age) <= 60, 
                                 "Возраст должен быть между 18 и 60 годами и целым числом"))

validate = is_id & is_name & is_age 


def insert(user: User, to_db: List[User]) -> None:
    is_valid, error = validate(user)
    if not is_valid:
        raise ValueError(error)
    
    to_db.append(user)
    
def show(db: List[User]) -> None:
    headers = ["ID", "Имя", "Фамилия", "Возраст"]
    print(" | ".join(f"{h:^10}" for h in headers))
    print("-" * 50)
    
    for user in sorted(db, key=lambda user: int(user.id)):
        row = [
            f"{user.id:^10}",
            f"{user.firstname:^10}",
            f"{user.lastname:^10}",
            f"{user.age:^10}"
        ]
        print(" | ".join(row))
    
################################################

db = []

user_input = [
    "иван башаров 3422ее 56!32", 
    "коля петров 30 932",
    "ВаЛЕнтиН сМехоВ 40 9",
    "стас Михайлов 59 932",
    "кристина бойко 25 1233345435",
    "ПАВЛОВА НАТАЛЬЯ 49 0000012"   
]

for input in user_input:
    try:
        firstname, lastname, age, id = input.split()

        insert(User(id, firstname, lastname, age), to_db=db)
    except Exception as ex:
        print(input, '->', ex)

print(end="\n\n")

show(db)
```
