1. Пользователь в бесконечном цикле вводит числа через консоль до тех пор, пока не введёт что-то, отличное от числа (после этого цикл завершается). В конце программа должна вывести сумму всех введённых чисел.


```python
user_input = ["234", "323", ".05", "-.333", "dfsd", "33333"]

summ = 0 
for number in user_input:
    try:
        summ += float(number)
    except ValueError:
        break

print(summ)
```


2. Пользователь вводит строку через консоль (состоящую только из букв в разном регистре). Программа должна создать новую строку, содержащую только те буквы из исходной строки, которые были в верхнем регистре.


```python
user_input = "dfJKJdnmmnfiwhuHUEHBNbwnJdhJhdwnmnbNBdh"

result = "".join(symbol for symbol in user_input if symbol.isupper())
print(result)
```


3. Классическая игра FizzBuzz. Программа в цикле выводит в консоль числа от 1 до 100.

Если число кратно 3, вместо него выводится Fizz.

Если кратно 5 — Buzz.

Если кратно и 3, и 5 — FizzBuzz.


```python
user_input = [1, 23, 55, 100, 36, 255]
dict_ = {3: "Fizz", 5: "Buzz"}

# Вариант 1
for input in user_input:
    print(input, '->', end=" ")    
    for key, value in dict_.items():
        if input % key == 0:
            print(value, end="")
    print()

# Вариант 2
for input_val in user_input:
    print(input_val, '->', end=" ")
    print("".join(value for key, value in dict_.items() if input_val % key == 0))
```

4. Пользователь вводит строку, содержащую любые символы (буквы, цифры и специальные символы). Программа должна вычислить и вывести сумму всех цифр, присутствующих в этой строке.

```python
user_input = "dkfj89f7uh32kdjfh89er32r09ed9f-232n423l9c!!!0923i042k"

result = sum(int(symbol) for symbol in user_input if symbol.isdigit())

print(result)
```


5. Пользователь вводит через консоль 10 чисел. Программа должна определить максимальное число и вывести его в консоль.

```python
user_input = [32, 44, 121, 555, 12, -40, -44, 11, 88]

print(max(user_input))
```
