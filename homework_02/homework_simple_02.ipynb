1. Написать программу, которая получает на вход строку и возвращает словарь, где:
ключи — символы из этой строки;

значения — количество раз, сколько каждый символ встречается.

```python
user_input = "fkjvnvo4h3p90230ruhsd8fb029!@!((!#&8-2437++++~__+120-293))))))"

symbols = {symbol: user_input.count(symbol) for symbol in sorted(set(user_input))}
print(symbols)
```


2. Дана строка текста (или введённая через консоль). Программа должна вернуть новую строку, в которой порядок слов будет обратным.
Пример:
"Python is really cool" → "cool really is Python".

```python
user_input = "Python is really cool" # → "cool really is Python"

print(" ".join(word for word in user_input.split()[::-1]))
```

4. Даны три (или больше) списка с объектами. Программа должна создать новый список, содержащий все уникальные элементы — каждый объект встречается только один раз.

```python
import itertools

user_input = [
    [3, 1, 6, 4, 66, "ff"],
    ["ewr", "3", "66", "ff"],
    ["!", 66, 12, ".4d"]   
]

print(list(set(itertools.chain(*user_input))))
```

5. Дана строка текста (или введённая через консоль). Программа должна вернуть словарь с четырьмя ключами:

"гласные",

"согласные",

"цифры",

"пунктуация".

Значения — количество символов каждого типа в строке.

```python
import string
from pprint import pprint

user_input = "ls;vjfvЩУШКГ№)шщаоц0у9кг32-чсъoioioioйьбтьоешщропмсьтмсоыо!!!,,,,БЮЮБЮЮююююЮЮ,.....в.ыаыац)№)@!!#%)^*((&_+))"

result = {
    "гласные": set(),
    "согласные": set(),
    "цифры": set(),
    "пунктуация": set(),
}

VOWELS = "аеёиоуыэюяaeiou"
CONSONANTS = "бвгджзйклмнпрстфхцчшщbcdfghjklmnpqrstvwxyz"

for symbol in user_input:
    if symbol.isdigit():
        result["цифры"].add(symbol)
    elif symbol in string.punctuation:
        result["пунктуация"].add(symbol)
    elif symbol.lower() in VOWELS:
        result["гласные"].add(symbol.lower())
    elif symbol.lower() in CONSONANTS:
        result["согласные"].add(symbol.lower())

result = {key: sorted(list(value)) for key, value in result.items()}

pprint(result, width=120, compact=True)
```
